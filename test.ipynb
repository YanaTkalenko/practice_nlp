```python
def factorial(number, level=0):
  indent = '    ' * level
  print(f'{indent}factorial({number})')

  if number <= 1:
        print(f'{indent}Возвращаем 1')
        return 1

  else:
    previous = factorial(number - 1, level + 1)
    result = number * previous

    print(f'{indent}{number} * {previous} = {result}')
    return result
```

```python
factorial(5)
```
